```python
sourcepath = "/Volumes/ev_spark/myvol/landing/postal_codes/"
```

## Method to read Postal Codes from source file

```python
def readPostalCodes():
        from pyspark.sql.types import StructType, StructField, StringType, DoubleType, IntegerType
        from pyspark.sql.functions import current_timestamp
        schema = StructType([
                StructField("CountryCode", StringType(), True),
                StructField("FSA", StringType(), True),
                StructField("City", StringType(), True),
                StructField("Province", StringType(), True),
                StructField("ProvinceCode", StringType(), True),
                StructField("admin_name2", StringType(), True),
                StructField("admin_code2", StringType(), True),
                StructField("admin_name3", StringType(), True),
                StructField("admin_code3", StringType(), True),
                StructField("Latitude", DoubleType(), True),
                StructField("Longitude", DoubleType(), True),
                StructField("Accuracy", IntegerType(), True)
        ])
        print("Reading from source")
        readPostalCodes_df = (spark.read
                        .format("csv")
                        .option("sep", "\t")
                        .option("header", "false")
                        .schema(schema)
                        .load(sourcepath)
                        .withColumn("extraction_date", current_timestamp())
                )
        print("Read Complete")
        print("*****************************")
        return readPostalCodes_df

```

## Method to write Postal Codes to bronze table

```python

def writePostalCodes(df):
    print("Writing to Bronze")
    (df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("ev_spark.bronze.postal_codes"))
    print("Write Complete")
    print("********************************")
```

## Invoking methods to read Postal Codes from source and write Postal Codes to Bronze

```python
readPostalCodes_df = readPostalCodes()
writePostalCodes(readPostalCodes_df)
```
