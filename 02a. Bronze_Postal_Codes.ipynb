```python
sourcepath = "/Volumes/ev_spark/myvol/landing/postal_codes/"
```

## Method to read Postal Codes from source file

```python
def readPostalCodes():
        print("Reading from source")
        readPostalCodes_df = (spark.read
                        .format("csv")
                        .option("sep", "\t")
                        .option("header", "false")
                        .option("inferSchema", "true")
                        .load(sourcepath)
                        .toDF("geoname_id",
                                "name",
                                "ascii_name",
                                "alternate_names",
                                "latitude",
                                "longitude",
                                "feature_class",
                                "feature_code",
                                "country_code",
                                "cc2",
                                "admin1_code",
                                "admin2_code",
                                "admin3_code",
                                "admin4_code",
                                "population",
                                "elevation",
                                "dem",
                                "timezone",
                                "modification_date"))
        print("Read Complete")
        print("*****************************")
        return readPostalCodes_df

```

## Method to write Postal Codes to bronze table

```python

def writePostalCodes(df):
    print("Writing to Bronze")
    (df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("ev_spark.bronze.postal_codes"))
    print("Write Complete")
    print("********************************")
```

## Invoking methods to read Postal Codes from source and write Postal Codes to Bronze

```python
readPostalCodes_df = readPostalCodes()
writePostalCodes(readPostalCodes_df)
```
