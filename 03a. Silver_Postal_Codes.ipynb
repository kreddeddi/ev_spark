```python
def readBronzeGeo():
    df_silver_postal = (spark.read
                            .table("ev_spark.bronze.postal_codes")
                            .select("CountryCode", "FSA", "City", "Province", "ProvinceCode","Latitude", "Longitude", "Accuracy", "extraction_date")
    )
    return df_silver_postal
```

```python
def writeSilverGeo(df):
    (df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable("ev_spark.silver.postal_codes")
    )
```

```python
df_geoBronze = readBronzeGeo()
writeSilverGeo(df_geoBronze)
```
